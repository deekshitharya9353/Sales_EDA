```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"]=(10,5)
```

```python
df['gender']=df['gender'].replace('Non-Binary','Others')
```

```python
count = (
    df.groupby('customer_age')['customer_id']
      .count()
      .sort_values(ascending=False)
      .head(10)
      .reset_index(name='customer_count')
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x='customer_age',
    y='customer_count',
    data=count,
    hue='customer_age',
    palette='Set3',
    legend=False
)

plt.title('Top 10 Customer Ages by Customer Count')
plt.xlabel('Customer Age')
plt.ylabel('Number of Customers')
```

```python
count = (
    df.groupby('customer_segment')['customer_id']
      .count()
      .sort_values(ascending=False)
      .head(10)
      .reset_index(name='customer_count')
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x='customer_segment',
    y='customer_count',
    data=count,
    hue='customer_segment',
    palette='Set3',
    legend=False
)

plt.title('Top 10 Customer Segment by Customer Count')
plt.xlabel('customer_segment')
plt.ylabel('Number of Customers')
```

```python
count = (
    df.groupby('customer_city')['customer_id']
      .count()
      .sort_values(ascending=False)
      .head(10)
      .reset_index(name='customer_count')
)

plt.figure(figsize=(12, 6))

sns.barplot(
    x='customer_city',
    y='customer_count',
    data=count,
    hue='customer_city',
    palette='Set3',
    legend=False
)

plt.title('Top 10 customer city by Customer Count')
plt.xlabel('customer_segment')
plt.ylabel('Number of Customers')
```

```python
count = (
    df.groupby('gender')['customer_id']
      .count()
      .sort_values(ascending=False)
      .head(10)
      .reset_index(name='customer_count')
)
plt.figure(figsize=(4,4))
plt.pie(count['customer_count'],labels=count['gender'],autopct='%1.1f%%',startangle=90,pctdistance=0.6,textprops={'fontsize':10})
```

```python
count = (
    df.groupby('region')['customer_id']
      .count()
      .sort_values(ascending=False)
      .head(10)
      .reset_index(name='customer_count')
)
plt.figure(figsize=(4,4))
plt.pie(count['customer_count'],labels=count['region'],autopct='%1.1f%%',startangle=90,pctdistance=0.6,textprops={'fontsize':10})
```

```python
count = (
    df.groupby(['customer_segment', 'customer_country'])['customer_id']
      .count()
      .reset_index(name='customer_count')
      .sort_values('customer_count', ascending=False)
)

plt.figure(figsize=(14, 7))

sns.barplot(
    data=count,
    x='customer_country',
    y='customer_count',
    hue='customer_segment'
)

plt.title('Customer Segment Count by Country')
plt.xlabel('Country')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.legend(title='Customer Segment')
plt.tight_layout()
plt.show()
```
